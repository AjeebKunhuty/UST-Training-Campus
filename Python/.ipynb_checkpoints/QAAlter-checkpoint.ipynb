```python
import pandas as pd

df = pd.read_csv('Dataset/AWS02.csv')
df
```

```python
df['Total'] = df.sum(axis=1, numeric_only=True)
df
```

```python
df['Percent'] = (df['Total']/27)*100
df
```

```python
df = df.reset_index()
percentCounts = df.groupby('Percent')['Total'].count()
percentCounts
```

```python
bins = [0,40,50,60,70,80,90,100]
labels = ['<40','<50','<60','<70','<80','<90','<100']

df['group'] = pd.cut(df['Percent'], bins = bins, labels = labels, right = False)

values = df.value_counts('group').sort_index(key = lambda x: x.str.replace('<','').astype(int))
values
```

```python
import matplotlib.pyplot as plt

plt.bar(labels, values, color = ['red', 'g', 'black'], width = 0.8)

plt.xlabel('Categories')
plt.ylabel('Count')
plt.title('Exam Result Division')

plt.show()
```

```python
df['Level'] = df['Percent'].apply( lambda x : 'Advanced' if x >= 70 else 'Beginner')
df
```

```python
df.value_counts('Level')
```
