```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
traffic = [150,200,180, 220, 250, 210]
```

```python
# plt.plot(months, traffic)
plt.plot(months, traffic, marker = '*', markersize = 15, color = 'blue', linestyle = '-.', linewidth = 2, label = "Traffic")
plt.xlabel('Month')
plt.ylabel('Monthly Traffic')
plt.title('Monthly Website Traffic')
# plt.legend()
plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
product_a_revenue = [45,55,40,70,80,75]
product_b_revenue = [35,40,50,55,75,68]

plt.plot(months, product_a_revenue, marker = 'o', linestyle = '-', color = 'blue', label = "Product A")
plt.plot(months, product_b_revenue, marker = 's', linestyle = '--', color = 'red', label = "Product B")

plt.xlabel('Month')
plt.ylabel('Monthly Revenue (in $1000)')
plt.title('Monthly Revenue Comparison')

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv(r"Dataset/IPL Matches 2008-2020.csv")
df.head(2)
```

```python
df[df['toss_winner'] == df['winner']]
```

```python
totalMatches = df.count()
totalMatches
```

```python
totalTossWinners = df[df['toss_winner'] == df['winner']].count()
totalTossWinners
```

```python
categories = ["Win", "Loss"]
wins = [totalTossWinners.id, totalMatches.id - totalTossWinners.id]
wins
```

```python
plt.bar(categories, wins, color = ['green', 'black'], width = 0.8)

plt.xlabel('Match Result')
plt.ylabel('Win Count')

plt.title("Match winner compared to Toss winner")
plt.show()
```
