```python
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv(r"Dataset/IPL Matches 2008-2020.csv")
df.head(2)
```

```python
df[df['toss_winner'] == df['winner']]
```

```python
totalMatches = df.count()
totalMatches
```

```python
totalTossWinners = df[df['toss_winner'] == df['winner']].count()
totalTossWinners
```

```python
categories = ["Win", "Loss"]
wins = [totalTossWinners.id, totalMatches.id - totalTossWinners.id]
wins
```

```python
plt.bar(categories, wins, color = ['green', 'black'], width = 0.8)

plt.xlabel('Match Result')
plt.ylabel('Win Count')

plt.title("Match winner compared to Toss winner")
plt.show()
```

```python
quarters = ['Q1', 'Q2', 'Q3', 'Q4']
electronics = [10000,12000,11000,10500]
clothing = [5000, 6000, 7500, 8000]
home_appliances = [3000, 7500, 8200, 9000]

plt.figure(figsize=(10,6))
plt.stackplot(quarters, electronics, clothing, home_appliances, 
              labels = ['Electronics','Clothing','Home Appliances'])

plt.xlabel('Quarters')
plt.ylabel('Sales ($)')
plt.title('Product Category Sales Over Quarters')
plt.legend()

plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
x, y = np.meshgrid(x,y)
z = np.sin(np.sqrt(x**2 + y**2))

fig = plt.figure()
ax = fig.add_subplot(111, projection = '3d')
ax.scatter(x,y,z,  c=z,cmap='viridis')
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')

plt.show()
```

```python
x = np.linspace(0,10,100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y4 = np.exp(-x)

fig, ax = plt.subplots(2,2,figsize=(10,8))

ax[0,0].plot(x,y1,'r')
ax[0,0].set_title('Sin')
ax[0,1].plot(x,y2,'b')
ax[0,1].set_title('Cos')
ax[1,0].plot(x,y3,'g')
ax[1,0].set_title('Tan')
ax[1,1].plot(x,y4,'m')
ax[1,1].set_title('Exponentail Decay')

plt.tight_layout()
plt.show()
```
