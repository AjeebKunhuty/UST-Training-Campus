```python
avgBP = df['trestbps'].mean()
avgBP
```

```python
highBP = df[df['trestbps'] > 140]
highBP
```

```python
diseased = highBP.condition.sum()
nondiseased = highBP.condition.count() - diseased
plt.pie([diseased, nondiseased] , labels = ['Diseased', 'Non-diseased'], autopct = "%2.2f%%")
plt.legend()
plt.title("Disease presence for patients with BP > 140")
plt.show()
```

```python
illThala = df[df['condition'] == 1].thalach
nonIllThala = df[df['condition'] == 0].thalach
print(illThala.mean(), nonIllThala.mean())
```

```python
plt.boxplot(list(illThala))
plt.show()
```

```python
exang = df[df['exang'] == 1]
nonExang = df[df['exang'] == 0]
exangCount = exang.exang.count()
nonExangCount = nonExang.exang.count()
exangPercent = (exang.condition.sum() / exangCount) *100
nonExangPercent = (nonExang.condition.sum() / nonExangCount) *100
print(f"Exang with disease percentage { exangPercent : .2f}")
print(f"Non-exang with disease percentage { nonExangPercent : .2f}")
```

```python
plt.bar(['Exang', 'Non-exang'],[exangPercent, nonExangPercent])
plt.ylim(0,100)
plt.show()
```

```python
illOldPeak = df[df['condition'] == 1].oldpeak
nonIllOldPeak = df[df['condition'] == 0].oldpeak


print("Diseased mean of ST Depression ",illOldPeak.mean())
print("Non-diseased mean of ST Depression ",nonIllOldPeak.mean())

plt.hist(illOldPeak, width = 0.2)
plt.title("ST Depression histogram of diseased")
plt.show()

plt.hist(nonIllOldPeak, width = 0.2)
plt.title("ST Depression histogram of non-diseased")
plt.show()
```

```python
caGroup = df.groupby('ca').condition
```

```python
caIll = caGroup.sum()
caAll = caGroup.count()
caProbability = caIll/caAll 
caProbability
```

```python
plt.plot(caProbability.index, caProbability)
plt.xlabel('Number of major vessels')
plt.ylabel('Disease probability')
plt.show()
```

```python
cross_tab = pd.crosstab(df['thal'], df['condition'], normalize='index') * 100 
cross_tab.plot(kind='bar', stacked=True,figsize=(10, 6))

plt.title('Thalassemia Type vs Disease Probability')
plt.xlabel('Thalassemia')
plt.ylabel('Percent')
plt.legend(['No Disease', 'Disease'], title='Condition', loc='upper right')
# plt.show()
# print(pd.crosstab.__doc__)
```

```python
condition = df['condition'].to_numpy()
patients = (df['age'].to_numpy() > 50) & (df['chol'].to_numpy() > 240) & (df['trestbps'].to_numpy() > 140)
diseasePercentage = np.mean(condition[patients]) * 100
diseasePercentage
```

```python
df['risk_score'] = (df['chol'] / 200) + (df['trestbps'] / 120) + df['oldpeak']
conditions = [
    (df['risk_score'] <= 2),
    (df['risk_score'] > 2) & (df['risk_score'] <= 4),
    (df['risk_score'] > 4)
]
risk_labels = ['Low Risk', 'Medium Risk', 'High Risk']
df['risk_category'] = np.select(conditions, risk_labels, default='Medium Risk')

plt.figure(figsize=(10, 6))
plt.hist(df['risk_score'], bins=20, edgecolor='black')
plt.title('Distribution of Risk Scores')
plt.xlabel('Risk Score')
plt.ylabel('Frequency')
plt.show()
print(df['risk_category'].value_counts())
```
